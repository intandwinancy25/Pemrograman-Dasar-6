C. Contoh Penggunaan If-Elif-Else (If Majemuk)

```python
nilai = 73
if nilai >= 85:
    print("Luar Biasa! Nilai A")
elif nilai >= 70:
    print("Bagus! Nilai B")
elif nilai >= 60:
    print("Cukup. Nilai C")
else:
    print("Nilai Kurang, Perlu Belajar Lebih Giat.")
```

Penjelasan:

Program ini digunakan untuk menentukan kategori nilai berdasarkan angka yang diperoleh. Nilai diperiksa secara berurutan, dimulai dari kondisi if, dilanjutkan ke elif, dan diakhiri dengan else jika tidak ada kondisi sebelumnya yang terpenuhi. Berdasarkan contoh pada PPT, nilai 85 ke atas mendapatkan kategori A, nilai 70 sampai kurang dari 85 mendapatkan kategori B, nilai 60 sampai kurang dari 70 mendapatkan kategori C, sedangkan nilai di bawah 60 dinyatakan masih perlu ditingkatkan. Dengan percabangan ini, program dapat memberikan hasil yang berbeda sesuai dengan nilai yang dimasukkan.

```python
# Menentukan Jenis Diskon Berdasarkan Hari
hari: str = "Minggu"

if hari == "Sabtu" or hari == "Minggu":
    diskon: float = 0.20 # Diskon 20%
    print(f"Selamat! Anda mendapat diskon akhir pekan sebesar {diskon*100}%.")
elif hari == "Jumat":
    diskon: float = 0.15 # Diskon 15%
    print(f"Diskon spesial Jumat! Anda mendapat potongan {diskon*100}%.")
else:
    diskon: float = 0.05 # Diskon 5%
    print(f"Anda mendapat diskon hari biasa sebesar {diskon*100}%.")
```

Penjelasan:

Program ini menentukan persentase diskon berdasarkan hari yang tersimpan dalam variabel hari. Jika hari adalah Sabtu atau Minggu, pelanggan mendapatkan diskon 20%. Jika hari adalah Jumat, diskon yang diberikan sebesar 15%. Sementara itu, pada hari lainnya, pelanggan mendapatkan diskon 5%. Kondisi diperiksa secara berurutan menggunakan if, elif, dan else, kemudian program menampilkan informasi diskon sesuai dengan hari yang ditentukan.

```python
Rupiah = int(input("Masukkan Jumlah Uang (Rp): "))
print("1. Dollar (USD)")
print("2. Euro (EUR)")
print("3. Yen (JPY)")
pilih = int(input("Pilih konversi (1/2/3): "))

if pilih == 1:
    print("Dalam Dollar:", Rupiah / 15000)
elif pilih == 2:
    print("Dalam Euro:", Rupiah / 16000)
elif pilih == 3:
    print("Dalam Yen:", Rupiah / 120)
else:
    print("Pilihan Tidak Valid")
```

Program ini digunakan untuk mengonversi jumlah uang dalam rupiah ke mata uang asing sesuai pilihan pengguna. Pertama, pengguna memasukkan jumlah uang dalam rupiah. Setelah itu, program menampilkan tiga pilihan mata uang, yaitu dolar Amerika (USD), euro (EUR), dan yen Jepang (JPY). Jika pengguna memilih angka 1, jumlah rupiah dibagi dengan 15.000 untuk memperoleh nilai dalam dolar. Pilihan angka 2 membagi jumlah rupiah dengan 16.000 untuk memperoleh nilai dalam euro, sedangkan pilihan angka 3 membaginya dengan 120 untuk memperoleh nilai dalam yen. Jika pilihan tidak sesuai dengan ketiga opsi tersebut, program menampilkan pesan bahwa pilihan tidak valid. Nilai pembagi dalam program merupakan angka yang sudah ditentukan pada contoh, sehingga hasilnya mengikuti asumsi konversi tersebut.

Soal 1 — Kalkulator Faktorial (1 sampai 5)

Buatlah program yang:
1. Meminta pengguna memasukkan sebuah bilangan bulat antara 1 sampai 5.
2. Menghitung faktorial dari bilangan tersebut menggunakan if / elif / else — bukan
dengan fungsi bawaan Python.
3. Menampilkan hasil faktorial.
4. Jika input di luar rentang 1–5, tampilkan pesan bahwa input tidak valid.


```python
# Jawaban Soal 1 - Kalkulator faktorial 1 sampai 5

n= int(input("Masukkan bilangan bulat antara 1 sampai 5: "))

# Validasi input
if n < 1 or n > 5:
    print("Input tidak valid. Masukkan bilangan antara 1 sampai 5")

# Perhitungan faktorial
elif n == 1:
    print("1! = 1")
elif n == 2:
    print(f"2! = {2 * 1}")
elif n == 3:
    print(f"3! = {3 * 2 * 1}")
elif n == 4:
    print(f"4! = {4 * 3 * 2 * 1}")
else:
    print(f"5! = {5 * 4 * 3 * 2 * 1}")
```

Soal 2 — Klasifikasi Kualitas Daya Berdasarkan Faktor Daya dan
Efisiensi Distribusi
Buatlah program yang:
1. Meminta faktor daya (nilai antara 0.0 hingga 1.0) dan efisiensi distribusi (%).
2. Menghitung skor kualitas daya dengan rumus: skor = faktor_daya * efisiensi.
3. Mengklasifikasikan kualitas daya berdasarkan tabel berikut:
Cuplikan layar 2026-10-18 005435.png

```python

# Jawaban Soal 2: Klasifikasi Kualitas Daya

faktor_daya = float(input("Masukkan faktor daya (0.0 - 1.0): "))
efisiensi = float(input("Masukkan efisiensi distribusi (%): "))

skor = faktor_daya * efisiensi

print(f"Skor kualitas daya: {skor:.2f}")

if skor >= 85 and faktor_daya >= 0.90:
    klasifikasi = "PRIMA — sistem dalam kondisi optimal"
elif skor >= 70 and faktor_daya >= 0.80:
    klasifikasi = "BAIK — sistem memadai, pantau rutin"
elif skor >= 50 or faktor_daya >= 0.75:
    klasifikasi = "CUKUP — ada ruang perbaikan"
else:
    klasifikasi = "BURUK — perlu perbaikan segera"

print(f"Faktor daya: {faktor_daya}, efisiensi {efisiensi}%")
print(f"klasifikasi: {klasifikasi}")
```

Soal 3 — Rekomendasi Mode Operasi PLTS Berdasarkan Iradiasi dan
Kapasitas Baterai
Buatlah program yang:
1. Meminta iradiasi matahari (W/m²) dan SoC baterai (%).
2. Menghitung estimasi daya panel dengan rumus: daya_panel = iradiasi * 0.18 * 10
(asumsi: luas panel 10 m², efisiensi 18%).
3. Menentukan mode operasi berdasarkan tabel berikut:
image.png

```python
# Jawaban Soal 3 - Rekomendasi Mode Operasi PLTS

iradiasi = float(input("Masukkan iradiasi matahari (W/m²): "))
soc = float(input("Masukkan SoC baterai (%): "))

daya_panel = iradiasi * 0.18 * 10

print(f"Estimasi daya panel: {daya_panel:.1f} W")

if iradiasi > 600 and soc < 80:
    mode = "CHARGING - prioritaskan pengisian baterai"
elif iradiasi > 600 and soc >= 80:
    mode = "GRID-EXPORT - ekspor kelebihan daya ke jaringan"
elif iradiasi > 200 and soc >= 20:
    mode = "NORMAL - operasi mandiri"
elif iradiasi <= 200 and soc >= 50:
    mode = "BATTERY-ONLY - andalkan baterai"
else:
    mode = "EMERGENCY - kurangi beban segera"

print(f"Iradiasi: {iradiasi} W/m², SoC: {soc}%")
print(f"Mode operasi: {mode}")
```
